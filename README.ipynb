# Competitor Analysis Chatbot

A research-grade chatbot designed for deep competitor analysis.  
It gathers, verifies, and synthesizes information from multiple sources (web search, news, vector databases) and produces evidence-backed insights.  
Intended for product managers, researchers, analysts, and competitive intelligence teams.

---

## Quick Summary

- **Purpose:** A Retrieval-Augmented Generation (RAG) framework that searches the web, fetches content, summarizes and verifies information, stores/retrieves vectors, and produces citation-backed competitor insights.
- **Users:** Developers setting up the project, and analysts requiring reproducible workflows for competitive intelligence.

---

## Key Features

- **Multi-source retrieval**
  - DuckDuckGo web search  
  - News scraping  
  - Vector DB lookups (Milvus/Zilliz)  
- **RAG pipeline with LLM synthesis**
  - Selectable LLM providers (OpenAI, Azure, Hugging Face)
- **Caching & session management (Redis)**
- **FastAPI backend** (Search, Summarization, Analysis, Verification)
- **Streamlit UI** (`streamplite_app/app.py`)

---

## Technology Stack

- Python 3.10+
- FastAPI
- Streamlit
- Redis
- Zilliz/Milvus Vector DB
- OpenAI / Azure / Hugging Face LLMs
- Libraries: langchain, openai, pymilvus, streamlit, fastapi, uvicorn, python-dotenv

## Setup Instructions (Windows / PowerShell)

### 1. Clone the repository
```powershell
git clone <repo-url>
cd competitor_analysis_chatbot
```

### 2. Create a virtual environment
```powershell
python -m venv .venv
.\.venv\Scripts\Activate.ps1
```

### 3. Install dependencies  
```powershell
pip install -r final_requirement.txt
```
### 4. How to install radis 
document url : https://docs.servicestack.net/install-redis-windows#option-1-install-redis-on-ubuntu-on-windows

### 4. Keep all configuration in config/yaml file

---

## Running the App

### **Option A — Streamlit UI**
```powershell
streamlit run streamplite_app/app.py
```

### **Option B — Custom wrapper (`uiapp2.py`)**
```python
import streamplite_app.app
```

Run:
```powershell
streamlit run uiapp2.py
```

### **Option C — Only FastAPI**
```powershell
uvicorn app.main:app --reload
```
